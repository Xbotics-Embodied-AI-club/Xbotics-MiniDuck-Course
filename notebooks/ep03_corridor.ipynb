```python
gamma, slip = 0.9, 0.1
pi = {"右": 0.8, "左": 0.2}
V = [0.0, 0.0, 0.0]
```

```python
for sweep in range(1, 101):
    new = []
    for s in range(3):
        v = 0.0
        for a, p_a in pi.items():
            moved = s + 1 if a == "右" else max(s - 1, 0)
            for s_next, p in ((moved, 1 - slip), (s, slip)):
                r = 1.0 if s_next == 3 else 0.0
                v += p_a * p * (r + (0.0 if s_next == 3 else gamma * V[s_next]))
        new.append(v)
    V = new
    if sweep in (1, 2, 5, 20, 100):
        print(sweep, [round(x, 2) for x in V])
```
